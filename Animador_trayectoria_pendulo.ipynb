```python
# Activar matplotlib inline
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML
from matplotlib import rcParams

rcParams['animation.embed_limit'] = 50.0 
# Leer datos
def leer_datos(nombre_archivo):
    datos = np.loadtxt(nombre_archivo)
    t = datos[:, 0]
    x = datos[:, 1]
    y = datos[:, 2]
    return t, x, y

# Función de animación
def animar(i):
    # Actualizar posición del punto
    punto.set_data([x[i]], [y[i]])
    # Actualizar la línea desde (0,0) hasta (x[i], y[i])
    linea.set_data([0, x[i]], [0, y[i]])
    # Actualizar texto de tiempo
    tiempo_texto.set_text(f"t = {t[i]:.2f}")
    return punto, linea, tiempo_texto

# MAIN
dpi=150
nombre_archivo = 'data_cartesianas'
t, x, y = leer_datos(nombre_archivo)

fig, ax = plt.subplots()
ax.set_xlim(min(x)-1, max(x)+1)
ax.set_ylim(min(y)-1, max(y)+1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Simulación de un Péndulo')

# Crear el punto (masa del péndulo)
punto, = ax.plot([], [], 'ro', markersize=8)

# Crear la línea que conecta (0,0) con el punto
linea, = ax.plot([], [], 'b-', linewidth=2)

# Texto para mostrar el tiempo
tiempo_texto = ax.text(0.05, 0.95, '', transform=ax.transAxes)

# Crear animación
ani = animation.FuncAnimation(
    fig, animar, frames=len(t), interval=10, blit=True
)

# Cerrar figura para evitar imagen estática
plt.close(fig)

# Mostrar animación
HTML(ani.to_jshtml())

ani.save('pendulo.mp4', writer='ffmpeg', fps=90)

print("se guardo correctamente")
```
